T4 x2 BF16 cognitive-runtime bring-up with measured CPU offload.


```python
import os, json, psutil, torch, subprocess
print(subprocess.check_output('nvidia-smi', shell=True, text=True))
print({'gpus': torch.cuda.device_count(), 'ram_gib': psutil.virtual_memory().total/2**30, 'available_gib': psutil.virtual_memory().available/2**30})
```

```python
MODEL_ID='huihui-ai/Huihui-Qwen3.8-27B-abliterated'
GPU_RESERVE_GIB=5.0
HOST_RESERVE_GIB=12.0
MODEL_ESTIMATE_GIB=51.7
gpu_total=sum(torch.cuda.get_device_properties(i).total_memory for i in range(torch.cuda.device_count()))/2**30
host_available=psutil.virtual_memory().available/2**30
offload=max(0.0, MODEL_ESTIMATE_GIB-max(0.0,gpu_total-GPU_RESERVE_GIB))
assert offload <= max(0.0,host_available-HOST_RESERVE_GIB), 'Insufficient measured host memory'
print({'model':MODEL_ID,'gpu_total_gib':gpu_total,'cpu_offload_start_gib':round(offload,2)})
```
